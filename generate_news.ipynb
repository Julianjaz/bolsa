El rango de fechas para generar las noticias es desde 2022 hasta la actualidad

### Funciones

```python
def crawling_news_date(date):
  import requests
  from bs4 import BeautifulSoup
  url = f'https://news.google.com/search?q=Tesla%20{date}&hl=en-US&gl=US&ceid=US%3Aen'
  response = requests.get(url)
  html_content = response.text
  soup = BeautifulSoup(html_content, 'html.parser')
  news = soup.find_all(class_='JtKRv')
  soup = BeautifulSoup(html_content, 'html.parser')
  dates = soup.find_all(class_='hvbAAd')
  return news, dates


def convert_element_to_text_and_date(news,dates):
  import pandas as pd
  textos = [elemento.text for elemento in news]

  # Crea un DataFrame con una sola columna llamada "texto"
  df_news = pd.DataFrame(columns=['new'])

  # Llena el DataFrame con los textos resultantes
  df_news['new'] = textos
  dates_news = [elemento.get('datetime') for elemento in dates]

  df_news['date'] = dates_news
  return df_news
```

### Generar lista de fechas

```python
from datetime import datetime, timedelta
inicio = datetime(2022, 1, 1)
fin = datetime.now()
fechas = []

fecha_actual = inicio
while fecha_actual <= fin:
    fechas.append(fecha_actual)
    fecha_actual += timedelta(days=1)
```

```python
NEW=[]
DATE=[]
for i in fechas:
  fecha_string = i.strftime('%Y-%m-%d')
  new,date = crawling_news_date(fecha_string)
  NEW.append(new)
  DATE.append(date)
```

```python
news_list = []
dates_list = []

for lista in NEW:
    news_list.extend(lista)

for lista in DATE:
    dates_list.extend(lista)
```

#### Se genera el dataframe con la listas consultadas de noticias y fechas

```python
df_n = convert_element_to_text_and_date(news_list,dates_list)
```

```python
df_n.to_parquet('news.parquet')
```
