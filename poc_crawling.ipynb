```python
import requests
from bs4 import BeautifulSoup
```

```python
dates_list = ['2023-01-01','2023-01-02']
```

```python
news = []
for i in dates_list:
    url = (f'https://news.google.com/search?q=tesla%20{i}&hl=es-419&gl=CO&ceid=CO%3Aes-419')
    response = requests.get(url)
    html_content = response.text
    soup = BeautifulSoup(html_content, 'html.parser')
    news.append(soup.find_all(class_='JtKRv'))
```
