```python
from datetime import datetime, timedelta

# Define la fecha de inicio del año 2023
inicio_2023 = datetime(2023, 1, 1)

# Define la fecha de fin del año 2023
fin_2023 = datetime(2023, 12, 31)

# Crea una lista para almacenar las fechas del 2023
fechas_2023 = []

# Itera sobre cada día del año 2023 y agrega la fecha a la lista
fecha_actual = inicio_2023
while fecha_actual <= fin_2023:
    fechas_2023.append(fecha_actual)
    fecha_actual += timedelta(days=1)
```

```python
def crawling_news_date(date):
  import requests
  from bs4 import BeautifulSoup
  url = f'https://news.google.com/search?q=Tesla%20{date}&hl=en-US&gl=US&ceid=US%3Aen'
  response = requests.get(url)
  html_content = response.text
  soup = BeautifulSoup(html_content, 'html.parser')
  news = soup.find_all(class_='JtKRv')
  soup = BeautifulSoup(html_content, 'html.parser')
  dates = soup.find_all(class_='hvbAAd')
  return news, dates
```

```python
def convert_element_to_text_and_date(news,dates):
  import pandas as pd
  textos = [elemento.text for elemento in news]

  # Crea un DataFrame con una sola columna llamada "texto"
  df_news = pd.DataFrame(columns=['new'])

  # Llena el DataFrame con los textos resultantes
  df_news['new'] = textos
  dates_news = [elemento.get('datetime') for elemento in dates]

  df_news['date'] = dates_news
  return df_news
```
