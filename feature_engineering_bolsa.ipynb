```python
def prepare_df(df):
  df.columns = df.iloc[0]
  df = df[1:]
  df.reset_index(drop=True, inplace=True)
  df['Close'] = df['Close'].astype(float)
  df['High'] = df['High'].astype(float)
  df['Low'] = df['Low'].astype(float)
  df['Open'] = df['Open'].astype(float)
  df['Volume'] = df['Volume'].astype(float)
  nuevo_df = df.iloc[:, [0]]  # Seleccionar la primera columna "date"
  nuevo_df['Close'] = df['Close']  # Seleccionar otra columna
  nuevo_df['High'] = df['High']  # Seleccionar otra columna
  nuevo_df['Low'] = df['Low']  # Seleccionar otra columna
  nuevo_df['Open'] = df['Open']  # Seleccionar otra columna
  nuevo_df['Volume'] = df['Volume']  # Seleccionar otra columna
  nuevo_df = nuevo_df.dropna()
  nuevo_df['Date'] = pd.to_datetime(nuevo_df['Date'])
# Establecer la columna "date" como el índice del DataFrame
  nuevo_df.set_index('Date', inplace=True)
  return nuevo_df
```
